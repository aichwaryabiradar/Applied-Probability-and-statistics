```python
#Birthday problem for multiple number of people 

import matplotlib.pyplot as plt

def birthday_problem(n):

    if n > 365:  
        return 0.0
    i=0
    p=1

    for i in range(n):
        p*= (365-i)/365                                       #probability that no two people have same bithdays
    
    return p


num= list(range(1,101))
no_match = [birthday_problem(n) for n in num]

match_p=[1-birthday_problem(n) for n in num]                  #probability that two people have same bithdays

plt.figure()
plt.plot(num, no_match,marker="o")
plt.plot(num, match_p,color='red',marker="o")


plt.xlabel("Number of People (n)")
plt.ylabel("Probability")
plt.title("Birthday Problem Probabilities")

plt.show()


print(birthday_problem(1))  
print(birthday_problem(3))
print(birthday_problem(5))
print(birthday_problem(10))
print(birthday_problem(15))
print(birthday_problem(20))
print(birthday_problem(23))
print(birthday_problem(366))
```
