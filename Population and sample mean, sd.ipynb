```python
import numpy as np
 
# Step 1: Generate 10,000 random samples from a normal distribution
np.random.seed(42)
 
population = np.random.normal(loc=50, scale=10, size=10000)
 
# Step 2: Calculate population mean and standard deviation
population_mean = np.mean(population)
population_std = np.std(population, ddof=0)
 
# Step 3: Randomly select 30 samples from the population
sample = np.random.choice(population, size=30, replace=False)
 
# Step 4: Calculate sample mean and sample standard deviation
sample_mean = np.mean(sample)
sample_std = np.std(sample, ddof=1)
 
# Step 5: Calculate standard error
standard_error = sample_std / np.sqrt(len(sample))
 
# Display results
print("Population Mean:", population_mean)
print("Population Standard Deviation:", population_std)
 
print("\nSelected 30 Samples:")
print(sample)
 
print("\nSample Mean:", sample_mean)
print("Sample Standard Deviation:", sample_std)
print("Standard Error:", standard_error)
```
