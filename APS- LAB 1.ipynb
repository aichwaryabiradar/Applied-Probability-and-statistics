```python
#Problem1: A fair coin is tossed 100 times.Calculate: 
#Number of heads  
#Number of tails 
#Experimental probability of Head  
#Experimental probability of Tail

import random

n=100                                      #No. of trials
outcome=["Head","Tail"]                    #head and tail are the 2 outcomes

head,tail=0,0

for i in range(n):
    toss= random.choice(outcome)          #Randomly generating the outcome
    if toss== "Head":
        head=head+1
    elif toss == "Tail":                
        tail=tail+1


print("Number of head:",head)
print("Number of tail:",tail)

Exper_prob_h=head/n                        #Experimental probability= Count / No. of trials
Exper_prob_t=tail/n

print("Experimental probability of Head:",Exper_prob_h)
print("Experimental probability of Tail:",Exper_prob_t)

```

```python
#Problem 2: check Effect of the number of trials 
#Repeat the coin-toss experiment for: n=10, 100, 1,000, 10,000 

import random
                                     
outcome=["Head","Tail"]                      #head and tail are the 2 outcomes

head,tail=0,0

def cal_probability(n):                      #Defined a function for easy use
    head,tail=0,0
    for i in range(n):
        toss= random.choice(outcome)
        if toss== "Head":
            head=head+1
        elif toss == "Tail":
            tail=tail+1

    print("Number of Trials:",n)
    print("Number of head:",head)
    print("Number of tail:",tail)

    Exper_prob_h=head/n                        #Experimental probability= Count / No. of trials
    Exper_prob_t=tail/n

    print("Experimental probability of Head:",Exper_prob_h)
    print("Experimental probability of Tail:",Exper_prob_t)
    print("\n")

cal_probability(10)
cal_probability(100)
cal_probability(1000)
cal_probability(10000)

```

```python
#4. Addition rule for disjoint events 
#A die is rolled 10,000 times. 
#Define: A=getting 1, B= getting 6, check the events are joint or disjoint  
#Verify: P(A∪B)=P(A)+P(B) 

import random

n=10000
count_a,count_b,count_ab=0,0,0

for i in range(n):
    roll= random.randint(1,6)

    if roll==1:                                #Event A: Getting 1 
        count_a+=1
    elif roll==6:                              #Event B: Getting 6 
        count_b+=1
    if roll==1 and roll==6:                    #Getting both 
        count_ab+=1

if count_ab==0:
    print("Events are disjoint")
else:
    print("Events are joint ")
    

pa=count_a/n                       #probability = count/ no. of trials 
pb=count_b/n
pab=count_ab/n
print("Probabilityof getting 1 :" ,pa)             
print("Probabilityof getting 6 :",pb)

print("The Union of event A and B is :",pa+pb-pab)
print("P(A)+P(B) :",pa+pb)

```

```python
#Addition rule for overlapping events 
#For one die roll, define: A=getting an even number A={2,4,6} and B=getting a number greater than 3 B={4,5,6} 
#Verify: P(A∪B)=P(A)+P(B)−P(A∩B) 

import random

n=10000
count_a,count_b,count_ab,count_aorb=0,0,0,0

for i in range(n):
    roll= random.randint(1,6)

    if roll%2==0:                          #Event A: Getting even number
        count_a+=1
    elif roll>3:                           #Event B: Getting number greater than 3
        count_b+=1
        
    if roll%2==0 and roll>3:
        count_ab+=1
    elif roll%2==0 or roll>3:
        count_aorb+=1


pa=count_a/n
pb=count_b/n
pab=count_ab/n
paorb=count_aorb/n
print("Probabilityof getting even number :" ,pa)
print("Probabilityof getting number greater than 3 :",pb)

print("The Union of event A and B is :",paorb)
print("P(A)+P(B) - p(A and B) :",pa+pb-pab)

if count_ab==0:                            #Checking if disjoint or joint
    print("Events are disjoint")
else:
    print("Events are joint ")
```

```python
#Multiplication rule for independent events 
#A coin is tossed and a die is rolled 10,000 times. 
#Define: 
#obtaining Head  
#obtaining an even number 
#Verify: P(A∩B)=P(A)P(B) 

import random

n = 10000
count_a = 0
count_b = 0
count_ab = 0

outcome = ["Head", "Tail"]

for i in range(n):
    roll = random.randint(1, 6)
    toss = random.choice(outcome)                                                     
    if toss == "Head":                                          # Event A: obtaining Head
        count_a += 1

    if roll % 2 == 0:                                           # Event B: obtaining an even number
        count_b += 1

    if toss == "Head" and roll % 2 == 0:                         # A ∩ B: Head AND even number
        count_ab += 1

pa = count_a / n
pb = count_b / n
pab = count_ab / n

print("Probability of getting Head:", pa)
print("Probability of getting even:", pb)
print("P(A ∩ B):", pab)
print("P(A) × P(B):", pa * pb)

if abs(pab - pa * pb) < 0.01:
    print("The events are independent.")
else:
    print("The events may not be independent.")
```
